# Carga de SQL

```python
import pandas as pd
import sqlite3

df_produccion = pd.read_csv("../data/processed/produccion_limpia.csv")
df_deforestacion = pd.read_csv("../data/processed/deforestacion_limpia.csv")

df_produccion.head()
```

## creamos la base de datos y cargamos las dos tablas

```python
conexion = sqlite3.connect("../database/trazabilidad.db")

df_produccion.to_sql("produccion", conexion, if_exists="replace", index=False)
df_deforestacion.to_sql("deforestacion", conexion, if_exists="replace", index=False)

print("Tablas creadas correctamente")
```

```python
query = """
SELECT *
FROM produccion
LIMIT 5;
"""

pd.read_sql(query, conexion)
```

```python
query = """
SELECT *
FROM deforestacion
LIMIT 5;
"""

pd.read_sql(query, conexion)
```
